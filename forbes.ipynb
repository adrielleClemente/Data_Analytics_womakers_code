```python
from pydataset import data
import pandas as pd
```

```python
# Verifico o se o df está em ordem
forbes = data('Forbes2000')
forbes.head()
```

```python
#1. Qual a empresa mais valiosa da lista da Forbes 2000?

# -> 1°. Conferi o valor mais alto com .max()
# -> 2°. Criei uma variável para armazenar o df organizado por 'marketvalue' de forma decrescente. Então, com o head() selecionei só a 1° empresa, a mais valiosa.

#forbes['marketvalue'].max()
mais_valiosa = forbes.sort_values('marketvalue', ascending = False).head(1)
mais_valiosa
```

```python
# 2. Construa uma tabela com as TOP 10 empresas mais lucrativas da lista.

# -> 1°. Para obter os maiores valores utilizei a função nlargest, assim atravéz da coluna 'profits' os valores são verificdos e armazenados na variável.

top_10 = forbes.nlargest(10, 'profits')
top_10
```

```python
# 3. Qual a média de valores de empresa das cinco categorias mais valiosas?

# -> 1°. Crio uma variável agrupando os dados por 'category', desta forma encontro a soma do valor de mercado das 5 categorias.
# -> 2°. Com o atributo `.index` isolei os nomes de cada categoria já filtrada no passo anterior.
# -> 3°. A variável forbes_filtrado armazena somente os dados das empresas pertencentes às 5 categorias isoladas.
# -> 4°. Como em forbes_filtrado o nome das empresas não se repentem, para tirar a média de valores de empresas usei a função .mean().

forbes_cat = forbes.groupby('category')['marketvalue'].sum().nlargest(5)
# forbes_cat

indice_cat = forbes_cat.index
# indice_cat

forbes_filtrado = forbes[forbes['category'].isin(indice_cat)]
# forbes_filtrado

media_geral_empresas = forbes_filtrado['marketvalue'].mean().round(2)

print(f'A média de valores de empresa das cinco categorias mais valiosas de Forbes2000 é: {media_geral_empresas}')
```
